```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("fifa_data.csv")
print(df)
```

```python
country_count = df["Nationality"].value_counts().index[0]
print(country_count)
```

###  Country with the Most Players insight
 1-The country with the highest number of players shows which nationis strongest in producing football talent

2-This means that country has good traing systems,leagues,and opportunities for players to grow.

```python
import matplotlib.pyplot as plt


country_count = df["Nationality"].value_counts().head(5)
country_count .plot(kind = "bar",color="green")
plt.xlabel("Country")
plt.ylabel("Number of players")
plt.title("Top 5 countries with most number of players")
plt.show()
```

### Top 5 Countries insight
we can see the chart a few countries dominate football player prodection.The top countries have much higher numbers compared to others Which shows that football talent is not equally distributed around the world

```python
df["Wage"]=df["Wage"].str.replace("€","").str.replace("K","").astype(float)
highest_salary = df.loc[df["Wage"].idxmax()]


print(highest_salary["Name"])
print(highest_salary["Wage"])
```

###  the highest salary player insight
The player with the highest salary is likekly one of the best and most famous players. saary depends on performance,popularity and club value.

```python
import matplotlib.pyplot as plt

plt.hist(df["Wage"],color = "purple",bins = 20)
plt.xlabel("Wage")
plt.ylabel("Players")
plt.title("Wage Distribution")
plt.show()
```

### Histogram Insight
#. Most players earn average or low level salaries.


#. Only a few players earn very high salaries.

This means  income in football is not equal and only top players get very high pay

```python
foot_count = df["Preferred Foot"].value_counts()
print(foot_count)
```

```python
import matplotlib.pyplot as plt

foot_count.plot(kind = "bar",color = "red")
plt.xlabel("Foot")
plt.ylabel("Number of Players")
plt.title("Preferred Foot Players")
plt.show()
```

### Preferred Foot insight
Most players using their right foot and few players use their left foot.This shows right foot dominance in football,left-footed players are fewer and sometimes more vluable because they are rare
